```python
from src.tokenizers.BPE import BytePairEncoding
from src.datasets.evals import eval_parsing

tokenizer = BytePairEncoding.from_file(paths.EXPERIMENTS_DIR / "1B-model/hr_en_bs_tokenizer_48000")
parsers = {
     "eu-mmlu-hr": eval_parsing.parse_eu_mmlu_hr,
    "eu-mmlu-en": eval_parsing.parse_eu_mmlu_en,
    "belebele-en": eval_parsing.parse_belebele_en,
    "belebele-hr": eval_parsing.parse_belebele_hr,
    "flores-bs-en": eval_parsing.parse_flores_bs_en,
    "flores-en-hr": eval_parsing.parse_flores_en_hr,
    "copa-hr": eval_parsing.parse_copa_hr,
    "copa-en": eval_parsing.parse_copa_en,
}
parse_fn = eval

for preset in presets:
    dataset = load_dataset(str(paths.DATA_DIR / presets[preset][2]))
    flat_dataset = [
        example
        for split in dataset.values()
        for example in split
    ]
    parser = parsers.get(preset)
    #print(parser(flat_dataset[0], tokenizer)[0])

    for idx in range(5):
        print(idx)
        print("")
        print("---QUESTION_---")
        print(tokenizer.decode(parser(flat_dataset[idx], tokenizer)[0], add_special=True))
        print("---ANSWERS---")
                
        for ans in parser(flat_dataset[idx], tokenizer)[1]:
            print(tokenizer.decode(ans, add_special=True))
```
