```python
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

eos_token_id = 2
datasets = [
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/classla-web-bs-v1/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/classla-web-bs-v2/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/classla-web-hr-v1/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/classla-web-hr-v2/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/dgt-en-hr-v2021/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/fineweb-2-bos/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/fineweb-2-hrv/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/fineweb-edu-10BT/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/fineweb-hq-20BT/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/macocu-bs-en-sent/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/macocu-en-hr-sent/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/riznica/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/settimes-bs-en-v2/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/settimes-en-hr-v2/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/wikipedia-bs-20231101/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/wikipedia-en-20231101/tokenized_shuffled.npy"),
    Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000/wikipedia-hr-20231101/tokenized_shuffled.npy"),
]


def segment_lengths(a, chunk_size=1_000_000_000):
    previous = -1

    for start in range(0, len(a), chunk_size):
        chunk = a[start:start + chunk_size]
        positions = np.flatnonzero(chunk == 2) + start

        if len(positions):
            lengths = np.diff(positions, prepend=previous)
            yield lengths
            previous = positions[-1]


for dataset in datasets:
    x = np.load(dataset, mmap_mode="r")
    print(dataset)
    for lengths in segment_lengths(x):

        counts = np.zeros(3, dtype=np.int64)
        prev = -1

        for start in range(0, len(x), 1_000_000_000):
            pos = np.flatnonzero(x[start:start+1_000_000_000] == 2) + start
            if len(pos):
                lengths = np.diff(pos, prepend=prev)
                counts += [
                    lengths[lengths < 1024].sum(),
                    lengths[(lengths >= 1024) & (lengths <= 2048)].sum(),
                    lengths[lengths > 2048].sum()
                ]
                prev = pos[-1]
            break
    print(counts)
    pct = 100 * counts / counts.sum()
    print(pct)

    plt.bar(['<1024', '1024–2048', '>2048'], pct)
    plt.ylabel('% of elements')
    plt.show()
```
