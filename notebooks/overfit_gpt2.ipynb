```python
"""
Instantiation and overfitting of GPT2
"""
import torch
from torch.optim import AdamW
from torch.nn import CrossEntropyLoss

from src.models.gpt2 import GPT2
```

```python
transformer = GPT2(
    vocab_size=50257,
    N=12, 
    d_model=768,
    d_ff=768*4)

transformer = transformer.cuda()
    
num_params = sum(p.numel() for p in transformer.parameters())
print(f"Number of parameters: {num_params}")
batch_size = 32
seq_len = 32

x = torch.randint(low=0, high=50257, size=(batch_size, seq_len), dtype=torch.long).cuda()

#no grad
with torch.no_grad():   
    output = transformer(x.cuda())

print("Output shape:", output.shape)
print("Output:", output)

#check output distribution with loss

loss = CrossEntropyLoss()
optimizer = AdamW(transformer.parameters())
l = loss(output.view(-1, 50257), x.view(-1))
print("Loss:", l.item())
#loss should be around 9.2 for random output and random target, which is log(50257) since the model is essentially guessing randomly.
print("Expected loss for random output and random target:", torch.log(torch.tensor(50257.0)).item())


n_epochs = 10000

for i in range(n_epochs):
    optimizer.zero_grad()
    output = transformer(x.cuda())
    l = loss(output.view(-1, 50257), x.view(-1))
    l.backward()
    optimizer.step()
    print(l)


```
