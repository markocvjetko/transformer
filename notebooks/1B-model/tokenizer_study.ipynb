```python
from src.tokenizers.BPE import BytePairEncoding
from datasets import load_dataset
from src.utils import paths


from datasets import load_from_disk
from src.tokenizers.pretokenizers import hf_whitespace_pretokenizer

tokenizer_32000 = paths.EXPERIMENTS_DIR / "1B-model/hr_en_bs_tokenizer_32000"
tokenizer_48000 = paths.EXPERIMENTS_DIR / "1B-model/hr_en_bs_tokenizer_48000"
tokenizer_64000 = paths.EXPERIMENTS_DIR / "1B-model/hr_en_bs_tokenizer_64000"

# Load tokenized Wikipedia datasets from their output directories
wiki_hr_dir = paths.DATA_DIR / "raw/wikipedia-hr-20231101"
wiki_bs_dir = paths.DATA_DIR / "raw/wikipedia-bs-20231101"
wiki_en_dir = paths.DATA_DIR / "raw/wikipedia-en-20231101"

flores_en_path = paths.DATA_DIR / "raw/flores-en" / "dev" / "eng_Latn.jsonl"
flores_hr_path = paths.DATA_DIR / "raw/flores-hr" / "dev" / "hrv_Latn.jsonl"
flores_bs_path = paths.DATA_DIR / "raw/flores-bs" / "dev" / "bos_Latn.jsonl"

# Load datasets using their respective paths
ds_wiki_hr = load_dataset(str(wiki_hr_dir), split="train")
ds_wiki_bs = load_dataset(str(wiki_bs_dir), split="train")
ds_wiki_en = load_dataset(str(wiki_en_dir), split="train")

# Take only the first 1000 examples for each dataset
ds_wiki_hr = ds_wiki_hr.select(range(min(200, len(ds_wiki_hr))))
ds_wiki_bs = ds_wiki_bs.select(range(min(200, len(ds_wiki_bs))))
ds_wiki_en = ds_wiki_en.select(range(min(200, len(ds_wiki_en))))

ds_flores_en = load_dataset('json', data_files=str(flores_en_path), split="train")
ds_flores_hr = load_dataset('json', data_files=str(flores_hr_path), split="train")
ds_flores_bs = load_dataset('json', data_files=str(flores_bs_path), split="train")
```

```python
for tokenizer in [tokenizer_32000, tokenizer_48000, tokenizer_64000]:
    print(f"tokenizer: {tokenizer}")
    tokenizer = BytePairEncoding.from_file(tokenizer)
    #for dataset in [ds_wiki_en, ds_wiki_hr, ds_wiki_bs]:
    for dataset in [ds_flores_en, ds_flores_hr, ds_flores_bs]:
        n_text = 0
        n_words = 0
        n_tokens = 0
        unique_tokens = set()
        for x in dataset:
            words = hf_whitespace_pretokenizer(x["text"])
            tokens = tokenizer.tokenize(x["text"])
            for token in tokens:
                unique_tokens.add(token)

            n_text += len(x["text"])
            n_words += len(words)
            n_tokens += len(tokens)
       

        print("Number of words:", n_words)
        print("Number of tokens:", n_tokens)

        fertility = n_tokens / n_words
        compression = n_text / n_tokens
        unique_tokens_used = len(unique_tokens) / tokenizer.vocab_size
        print("Fertility (tokens per word):", fertility)
        print("compression (chars per token):", compression)
        print("Unique tokens used (fraction of vocab):", unique_tokens_used)
   
```
