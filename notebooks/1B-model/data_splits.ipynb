```python
from src.utils import paths
import os
import numpy as np
from src.tokenizers.BPE import BytePairEncoding

from pathlib import Path
root_dir = Path("/media/mcvjetko/marko/transformer/data/processed/1B-model-vocab-48000")

tokenizer_path = paths.EXPERIMENTS_DIR / "1B-model/hr_en_bs_tokenizer_48000"

datasets = {
    "classla-web-hr-v1":     {"path": root_dir / "classla-web-hr-v1" / "tokenized.npy",     "lang": "hr"},
    "classla-web-hr-v2":     {"path": root_dir / "classla-web-hr-v2" / "tokenized.npy",     "lang": "hr"},
    "fineweb-2-hrv":         {"path": root_dir / "fineweb-2-hrv" / "tokenized.npy",         "lang": "hr"},
    "riznica":               {"path": root_dir / "riznica" / "tokenized.npy",               "lang": "hr"},
    "classla-web-bs-v1":     {"path": root_dir / "classla-web-bs-v1" / "tokenized.npy",     "lang": "bs"},
    "classla-web-bs-v2":     {"path": root_dir / "classla-web-bs-v2" / "tokenized.npy",     "lang": "bs"},
    "fineweb-2-bos":         {"path": root_dir / "fineweb-2-bos" / "tokenized.npy",         "lang": "bs"},
    "fineweb-edu-10BT":      {"path": root_dir / "fineweb-edu-10BT" / "tokenized.npy",      "lang": "en"},
    "fineweb-hq-20BT":       {"path": root_dir / "fineweb-hq-20BT" / "tokenized.npy",       "lang": "en"},
    "macocu-en-hr-sent":     {"path": root_dir / "macocu-en-hr-sent" / "tokenized.npy",     "lang": "en-hr"},
    "macocu-bs-en-sent":     {"path": root_dir / "macocu-bs-en-sent" / "tokenized.npy",     "lang": "bs-en"},
    "settimes-en-hr-v2":     {"path": root_dir / "settimes-en-hr-v2" / "tokenized.npy",     "lang": "en-hr"},
    "settimes-bs-en-v2":     {"path": root_dir / "settimes-bs-en-v2" / "tokenized.npy",     "lang": "bs-en"},
    "dgt-en-hr-v2021":       {"path": root_dir / "dgt-en-hr-v2021" / "tokenized.npy",       "lang": "en-hr"},
}

wiki_datasets = {
    "wikipedia-hr":          {"path": root_dir / "wikipedia-hr-20231101" / "tokenized.npy",  "lang": "hr"},
    "wikipedia-bs":          {"path": root_dir / "wikipedia-bs-20231101" / "tokenized.npy",  "lang": "bs"},
    "wikipedia-en":          {"path": root_dir / "wikipedia-en-20231101" / "tokenized.npy",  "lang": "en"},
}

# Count total tokens per-dataset and per-language
tokens_per_lang = {}
tokens_per_dataset = {}

total_tokens = 0
for name, info in datasets.items():
    n = np.load(info["path"], mmap_mode="r").size
    tokens_per_dataset[name] = n
    tokens_per_lang[info["lang"]] = tokens_per_lang.get(info["lang"], 0) + n
    total_tokens += n

print("\nTotal tokens across all datasets: {:,}".format(total_tokens))

print("\nTotal tokens per language:")
for lang, toks in tokens_per_lang.items():
    print(f"{lang}: {toks:,}")

# Per-language normalization (fraction of total tokens)
normalized_tokens_per_lang = {
    lang: toks / total_tokens for lang, toks in tokens_per_lang.items()
}
print("\nNormalized tokens per language (fraction of total tokens):")
for lang, frac in normalized_tokens_per_lang.items():
    print(f"{lang}: {frac:.4%}")

# Per-dataset normalization (fraction of total tokens)
normalized_tokens_per_dataset = {
    name: toks / total_tokens for name, toks in tokens_per_dataset.items()
}
print("\nNormalized tokens per dataset (fraction of total tokens):")
for name, frac in normalized_tokens_per_dataset.items():
    print(f"{name}: {frac:.4%}")
```

```python
# Load the tokenizer
tokenizer = BytePairEncoding.from_file(tokenizer_path)

print("\nDecoding first 128 tokens from each dataset:")

for name, info in datasets.items():
    tokens = np.load(info["path"], mmap_mode="r")[:1024]
    # Ensure tokens are a list/array of ints
    if isinstance(tokens, np.ndarray):
        tokens = tokens.tolist()
    decoded = tokenizer.decode(tokens, add_special=True)
    print(f"\nDataset: {name}")
    print("\n\n\n\n\n\n")
    print(decoded)
```
